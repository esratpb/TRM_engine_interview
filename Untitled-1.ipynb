```python
%pip install scipy
import math
from scipy.stats import norm


def black_scholes_call(S, K, r, sigma, T):
    d1 = (math.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * math.sqrt(T))
    d2 = d1 - sigma * math.sqrt(T)

    call_price = S * norm.cdf(d1) - K * math.exp(-r * T) * norm.cdf(d2)

    return d1, d2, call_price
```

```python
# =========================
# Historical VaR Calculation
# =========================

import numpy as np

# Step 1: Total PnL already computed from dataframe
total_pnl = df_market_rate["Total PnL FXTEST Total PnL vector"].dropna()

# Step 2: Sort PnL (ascending: worst → best)
sorted_pnl = np.sort(total_pnl)

# Step 3: Number of observations
n = len(sorted_pnl)

# Step 4: Compute percentile index (1% for 99% VaR)
percentile_index = 0.01 * n

# Step 5: Adjust for Python 0-based indexing
lower_index = int(np.floor(percentile_index)) - 1
upper_index = int(np.ceil(percentile_index)) - 1

# Step 6: Interpolation weight
weight = percentile_index - np.floor(percentile_index)

# Step 7: Interpolated VaR
var = (
    (1 - weight) * sorted_pnl[lower_index] +
    weight * sorted_pnl[upper_index]
)

print("VaR (1-day, 99%):", var)
```

```python
import math
from scipy.stats import norm

def black_scholes(S, K, r, sigma, T, option_type="call"):
    d1 = (math.log(S / K) + (r + 0.5 * sigma**2) * T) / (sigma * math.sqrt(T))
    d2 = d1 - sigma * math.sqrt(T)

    if option_type == "call":
        return S * norm.cdf(d1) - K * math.exp(-r * T) * norm.cdf(d2)
    else:
        return K * math.exp(-r * T) * norm.cdf(-d2) - S * norm.cdf(-d1)
```

```python
# ITM
print(black_scholes(120, 100, 0.01, 0.3, 1, "call"))

# ATM
print(black_scholes(100, 100, 0.01, 0.3, 1, "call"))

# OTM
print(black_scholes(80, 100, 0.01, 0.3, 1, "call"))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# 1) d1 ve d2'yi TANIMLA
d1 = 0.65953
d2 = 0.45600

# 2) Normal dağılım eğrisi
x = np.linspace(-3, 3, 1000)
y = norm.pdf(x)

plt.figure(figsize=(8,4))
plt.plot(x, y, label="Standart Normal PDF")

# 3) d1 ve d2 çizgileri
plt.axvline(d1, color="dodgerblue", linestyle="--", label=f"d1 = {d1}")
plt.axvline(d2, color="orange", linestyle="--", label=f"d2 = {d2}")

plt.title("Normal Dağılım Üzerinde d1 ve d2")
plt.xlabel("Z")
plt.ylabel("Yoğunluk (PDF)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Black-Scholes parametreleri
S0 = 100      # Spot fiyat
K  = 100      # Strike
r  = 0.05     # Risksiz faiz (yıllık)
sigma = 0.20  # Volatilite (yıllık)
T  = 1.0      # Vade (yıl)

# 1) d1 ve d2 hesapla
d1 = (np.log(S0 / K) + (r + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
d2 = d1 - sigma * np.sqrt(T)

print("d1 =", d1)
print("d2 =", d2)

# 2) Normal dağılım eğrisi
x = np.linspace(-3, 3, 1000)
pdf = norm.pdf(x)

plt.figure(figsize=(8,4))
plt.plot(x, pdf, label="Standart Normal PDF")

# 3) d1 ve d2 çizgileri
plt.axvline(d1, color="dodgerblue", linestyle="--", label=f"d1 = {d1:.3f}")
plt.axvline(d2, color="orange", linestyle="--", label=f"d2 = {d2:.3f}")

plt.title("Black-Scholes: Normal Dağılım Üzerinde d1 ve d2")
plt.xlabel("Z")
plt.ylabel("Yoğunluk (PDF)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
